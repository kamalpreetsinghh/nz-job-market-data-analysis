## Initial observations

- Columns 1–10 contain regional indices.
- Columns 11–20 contain industry indices.

- Cleaning will be performed on copies of the raw data.

## Cleaning the regional data

Extract the date and regional indices into a separate DataFrame, preserving the original raw data.

```python
regional_raw = quarterly_raw.iloc[:, :11].copy()

regional_raw.shape
```

```python
regional_raw.head()
```

```python
regional_raw.columns = regional_raw.iloc[0]
```

```python
regional_raw.head()
```

```python
regional_clean = regional_raw.iloc[1:].copy()
regional_clean = regional_clean.dropna(how="all")
```

```python
regional_clean.head(10)
```

```python
region_columns = regional_clean.columns.drop("Date")
no_regional_data = regional_clean[region_columns].isna().all(axis=1)
regional_clean.loc[no_regional_data, ["Date"]]
```

```python
print("Rows with no regional data:", no_regional_data.sum())
```

```python
regional_clean = regional_clean.loc[~no_regional_data].copy()
regional_clean = regional_clean.reset_index(drop=True)
```

```python
regional_clean.head()
```

```python
regional_clean.isna().sum()
```

```python
regional_clean.columns.name = None
```

```python
regional_clean.dtypes
```

```python
regional_clean["Date"] = pd.to_datetime(
    regional_clean["Date"],
    errors="raise"
)
```

```python
regional_clean[region_columns] = regional_clean[region_columns].apply(
    pd.to_numeric,
    errors="raise"
)
```

```python
regional_clean.head()
```

```python
quarters = regional_clean["Date"].dt.to_period("Q")
```

```python
print("Rows", len(regional_clean))
print("First Quarter", quarters.min())
print("Last Quarter", quarters.max())
print("Duplicate quarters:", quarters.duplicated().sum())
print("Chronological Order", quarters.is_monotonic_increasing)
```

```python
expected_quarters = pd.period_range(
    start=quarters.min(),
    end=quarters.max(),
    freq="Q"
)

missing_quarters = expected_quarters.difference(
    pd.PeriodIndex(quarters)
)

print("Missing quarters:", missing_quarters.tolist())
```

## Export the cleaned regional data

```python
output_file = Path("../data/processed/regional_quarterly.csv")
regional_clean.to_csv(output_file, index=False)

print("File saved:", output_file.exists())
```

## Verify that the exported file loads correctly:

```python
regional_check = pd.read_csv(output_file, parse_dates=["Date"])

pd.testing.assert_frame_equal(
    regional_clean.reset_index(drop=True),
    regional_check,
    check_dtype=False,
    check_column_type=False
)

print("Export verification passed")
```
