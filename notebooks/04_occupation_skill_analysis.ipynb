# Occupation and Skill-Level Analysis

Compare annual changes in occupation and skill-level advertisement
indices. Detailed skill categories and broad skill groupings are
analysed separately. Index values are not vacancy counts.

```python
from pathlib import Path

import pandas as pd

processed_dir = Path("../data/processed")

analysis_files = {
    "Occupation": "occupation_quarterly.csv",
    "Detailed skill": "skill_levels_quarterly.csv",
    "Broad skill and overall": "broad_skill_quarterly.csv"
}

analysis_tables = {}

for name, filename in analysis_files.items():
    table = pd.read_csv(
        processed_dir / filename,
        parse_dates=["Date"],
        index_col="Date"
    )

    analysis_tables[name] = table
```

```python
annual_growth_tables = {}

for name, table in analysis_tables.items():
    growth = table.pct_change(
        periods=4,
        fill_method=None
    ) * 100

    annual_growth_tables[name] = growth

    print(name, "—", growth.index[-1].to_period("Q"))
    display(growth.iloc[-1].sort_values(ascending=False).round(2))
    print("------------------------")
```

```python
latest_broad_growth = (
    annual_growth_tables["Broad skill and overall"].iloc[-1]
)

broad_comparison = (
    latest_broad_growth[["Skilled", "Unskilled"]]
    .rename("yoy_change_pct")
    .to_frame()
)

broad_comparison["difference_from_overall_pp"] = (
    broad_comparison["yoy_change_pct"]
    - latest_broad_growth["ALL"]
)

broad_comparison.round(2)
```

### Conclusion: Occupation and skill-level growth

In 2026Q2, seven of eight occupation groups recorded positive annual
growth. Machinery Drivers led at 19.84%, while Community & Personal
Services declined by 0.18%.

Four of five detailed skill categories grew. The `unskilled` category
led at 12.12%, while the detailed `skilled` category declined by 0.30%.

Broad `Unskilled` growth was 10.21%, exceeding overall growth of 6.72%
by 3.49 percentage points. Broad `Skilled` grew by 5.51%.

Detailed categories and broad groupings are not interchangeable.
These results measure changes in advertisement indices, not
advertisement counts or employment shares.

### Classification reference

MBIE's 2018 Jobs Online methodology groups highly skilled, skilled,
and semi-skilled categories into broad `Skilled`, and low skilled
and unskilled categories into broad `Unskilled`.

[MBIE Jobs Online methodology (2018)](https://www.mbie.govt.nz/assets/82f9a170cc/jobs-online-methodology-2018.pdf)

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(15, 6))

chart_groups = [
    ("Occupation", "Occupation groups"),
    ("Detailed skill", "Detailed skill levels")
]

for ax, (group, title) in zip(axes, chart_groups):
    latest_growth = (
        annual_growth_tables[group]
        .iloc[-1]
        .sort_values()
    )

    colors = [
        "steelblue" if value >= 0 else "firebrick"
        for value in latest_growth
    ]

    bars = ax.barh(
        latest_growth.index,
        latest_growth.values,
        color=colors
    )

    ax.bar_label(
        bars,
        labels=[f"{value:+.2f}%" for value in latest_growth],
        padding=5
    )

    ax.set_title(title)
    ax.set_xlabel("Year-over-year change (%)")
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_xlim(-3, 23)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

latest_quarter = (
    annual_growth_tables["Occupation"].index[-1].to_period("Q")
)

fig.suptitle(
    f"Job-advertisement annual growth — {latest_quarter}",
    fontsize=14
)

fig.text(
    0.01,
    0.02,
    "Source: MBIE Jobs Online, June 2026 quarterly workbook. "
    "Annual changes calculated from unadjusted indices.",
    fontsize=9,
    color="dimgray"
)

fig.tight_layout(rect=[0, 0.07, 1, 0.95])
plt.show()
```
